## 一、到底什么是"机器学习"？

**大白话版：**

> 机器学习就是——**不直接告诉计算机规则，而是给它一堆例子，让它自己总结出规律。**

**生活类比：教小孩认猫**

- 传统编程的思路：你写一本《认猫手册》——"有胡须、尖耳朵、会喵喵叫……"。但规则**永远写不全**（折耳的猫呢？没尾巴的猫呢？）。
- 机器学习的思路：你给小孩看 1000 张猫的照片，每次说"这是猫"，看着看着，他自己**就"有感觉"了**。

机器学习干的就是后一件事：**从例子里自己学会规律。**

**吴恩达在课里给了两个定义，我们逐个拆：**

**① Arthur Samuel（1959）——通俗版：**
> 机器学习是让计算机**在没有被明确编程的情况下**也能学习的研究领域。

**② Tom Mitchell（1998）——正式版：**
> 一个程序如果随着**经验 E** 的增多，在**任务 T** 上的**表现 P** 越来越好，我们就说它在"学习"。

把 T / E / P 拆开看：

| 缩写 | 英文 | 中文意思 | 问自己一句 |
|---|---|---|---|
| **T** | Task | 任务 | 你要它干嘛？ |
| **E** | Experience | 经验 | 它看过多少例子？ |
| **P** | Performance | 表现 | 怎么判断它做得好不好？ |

**举个例子：垃圾邮件过滤器**
- T（任务）= 把每封邮件分成"垃圾 / 正常"
- E（经验）= 1 万封**已经人工标好**的邮件
- P（表现）= 它判断正确的百分比

你看，**P 会随着 E 变多而提高**——这就是"学习"。
